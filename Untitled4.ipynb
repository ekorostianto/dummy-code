```r
library(raster)
library(rgdal)
library(rasterVis)
library(RColorBrewer)

# set working directory to ensure R can find the file we wish to import
wd <- "D:/College/Semester8/data/analisa/grace_rl05/selected" # this will depend on your local environment environment
# be sure that the downloaded file is in this directory
#setwd(wd)

# Create list of NDVI file paths
#all_EWHGRACE <- list.files(paste0(wd,"grace_rl05/gracerl05_grd/4dan10"), full.names = TRUE, pattern = ".grd$")
all_EWHGRACE <- list.files(wd, full.names = TRUE, pattern = ".tif$")


# Create a time series raster stack
EWHGRACE_stack <- stack(all_EWHGRACE)

batasbasin <- readOGR(dsn=path.expand("D:/College/Semester8/data/das"),layer="DasBrantasBengawansolo")

masked <- mask(x = EWHGRACE_stack, mask = batasbasin)
cropped <- crop(x = masked, y = extent(batasbasin))

rasterNames  <- gsub("gracegrd_","", names(EWHGRACE_stack))
rasterNames  <- gsub(".csv","",rasterNames)

cols <- colorRampPalette(brewer.pal(9,"RdBu"))
#levelplot(cropped, nc = 12, main="Anomali EWH GRACE periode 2005-2015 (mm) \n Basin Bengawan Solo dan Brantas",
#xlab = "Longitude",ylab = "Latitude", col.regions=cols,names.attr=rasterNames) + layer(sp.polygons(batasbasin, col="black", border="black", lwd=1))
png("name.png")
print(levelplot(cropped, nc = 4, main="Anomali EWH GRACE periode 2005-2015 (mm) \n Basin Bengawan Solo dan Brantas",
xlab = "Longitude",ylab = "Latitude", col.regions=cols,names.attr=rasterNames) + layer(sp.polygons(batasbasin, col="black", border="black", lwd=1)))
dev.off()
```
