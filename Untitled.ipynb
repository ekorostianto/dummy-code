```r
library(rgdal)  
library(raster)
library(viridis)
wd <- "D:/College/Semester8/data/analisa/gws/monthly/" # this will depend on your local environment environment
# be sure that the downloaded file is in this directory
setwd(wd)
# import raster
GWSGLDAS <- raster(paste0(wd,"GIOVANNI-timeAvgMap.GLDAS_CLSM025_DA1_D_2_2_GWS_tavg.20150401-20150430.180W_60S_180E_90N.tif"))
batasbasin <- readOGR(dsn=path.expand("D:/College/Semester8/data/das"),
                            layer="DasBrantasBengawansolo")
masked <- mask(x = GWSGLDAS, mask = batasbasin)
cropped <- crop(x = masked, y = extent(batasbasin))
plot(cropped,
     main="Anomali Groundwater Storage April 2015 (mm) \n Basin Bengawan Solo dan Brantas",
     xlab = "Longitude", 
     ylab = "Latitude", 
     col=viridis(256))
plot(batasbasin, col="transparent", border="black", lwd=1, add = TRUE)
linesungai <- readOGR(dsn=path.expand("D:/College/Semester8/data/das"),
                            layer="linebengawansolobrantas")
plot(linesungai, col="blue", border="black", lwd=1, add = TRUE)
```
