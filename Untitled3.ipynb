```r
library(raster)
library(rgdal)
library(viridis)

ff <- list.files("D:/College/Semester8/data/analisa/gws/monthly", pattern = "\\.tif$", full=TRUE)
s <- stack(ff)
jumlah <- calc(s, sum)
vtahun <- (jumlah/length(ff))*12

library(rgdal)
batasbasin <- readOGR(dsn=path.expand("D:/College/Semester8/data/das"),
                            layer="DasBrantasBengawansolo")
masked <- mask(x = vtahun, mask = batasbasin)
cropped <- crop(x = masked, y = extent(batasbasin))
plot(cropped,
     main="Kecepatan rata2 per tahun EWH GRACE (mm) \n Basin Bengawan Solo dan Brantas",
     xlab = "Longitude", 
     ylab = "Latitude", 
     col=plasma(256)
    )
plot(batasbasin, col="transparent", border="black", lwd=1, add = TRUE)
linesungai <- readOGR(dsn=path.expand("D:/College/Semester8/data/das"),
                            layer="linebengawansolobrantas")
plot(linesungai, col="blue", border="black", lwd=1, add = TRUE)
```
